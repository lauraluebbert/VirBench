# Debug: gget virus Failure Modes (GPT 5.5)

Run `gget.virus()` with the exact filters from the 4 queries that GPT 5.5 consistently gets wrong even with gget:

| Query | Pathogen | Expected | GPT 5.5 returned | Delta |
|-------|----------|----------|-------------------|-------|
| Q3 | CCHFV (Africa) | 23 | 22 | -1 |
| Q36 | CCHFV (all) | 126 | 125 | -1 |
| Q69 | Dengue virus 2 (partial) | 2694 | 2907 | +213 |
| Q96 | Rabies lyssavirus (mouse) | 12 | 11 | -1 |

```python
import subprocess
# subprocess.run(["pip3", "install", "gget==0.30.3"], capture_output=True)

import gget
print(f"gget version: {gget.__version__}")
```

## Q3: CCHFV — Africa, 1000-2000 bp, max 10 N's (expected: 23, GPT got: 22)

```python
gget.virus(
    "3052518",
    geographic_location="Africa",
    min_collection_date="2020-01-01",
    max_collection_date="2025-12-31",
    min_release_date="2020-01-01",
    max_release_date="2025-12-31",
    min_seq_length=1000,
    max_seq_length=2000,
    max_ambiguous_chars=10,
    outfolder="debug_gget_q3",
)
```

```python
from pathlib import Path

fasta = list(Path("debug_gget_q3").glob("*.fasta"))
if fasta:
    count = sum(1 for line in open(fasta[0]) if line.startswith(">"))
    print(f"Q3 sequence count: {count} (expected: 23)")
else:
    print("No FASTA file found")

summary = Path("debug_gget_q3/command_summary.txt")
if summary.exists():
    print("\n" + summary.read_text())
```
